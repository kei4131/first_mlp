```python
# requires_grad
import torch

x = torch.tensor(2.0, requires_grad=True)

y = x ** 2

y.backward()

print(x.grad)

print(y.grad_fn) #rad_fn 就是在告诉你：这个 Tensor 是通过什么运算产生的，以及反向传播时该怎么把梯度传回去。
```

```python
import torch
import torch.nn as nn


x = torch.tensor([
    [1.],
    [2.],
    [3.],
    [4.]
])

target = torch.tensor([
    [3.],
    [5.],
    [7.],
    [9.]
])

model = nn.Linear(1, 1)
loss_fn = nn.MSELoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)

print("训练前：")
print(model.weight)
print(model.bias)

for i in range(1000):
    optimizer.zero_grad()

    pred = model(x)

    loss = loss_fn(pred, target)

    loss.backward()

    optimizer.step()

    if i % 10 == 0:
        print(i, loss.item())   # loss.item()把只包含一个元素的 Tensor 转成 Python 数字


print("训练后：")
print(model.weight)
print(model.bias)
```

```python
# dataset和dataloader
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader


# 1. Dataset
class MyDataset(Dataset):
    def __init__(self):
        self.x = torch.tensor([
            [1.],
            [2.],
            [3.],
            [4.]
        ])

        self.y = torch.tensor([
            [3.],
            [5.],
            [7.],
            [9.]
        ])

    def __len__(self):
        return len(self.x)

    def __getitem__(self, index):
        return self.x[index], self.y[index]


dataset = MyDataset()


# 2. DataLoader
loader = DataLoader(
    dataset,
    batch_size=2,
    shuffle=True
)


# 3. Model
model = nn.Linear(1, 1)

loss_fn = nn.MSELoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.01
)


# 4. Training
for epoch in range(100):

    for x_batch, y_batch in loader:

        optimizer.zero_grad()

        pred = model(x_batch)

        loss = loss_fn(pred, y_batch)

        loss.backward()

        optimizer.step()

    if epoch % 10 == 0:
        print(epoch, loss.item())
```
